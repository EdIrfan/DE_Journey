## **BOILER PLATE**

```python
import os
from dotenv import load_dotenv
from langchain_openrouter import ChatOpenRouter

load_dotenv(dotenv_path="../.env",override=True)

api_key=os.getenv("OPENROUTER_API_KEY")
if not api_key:
    raise RuntimeError("OPENROUTER_API_KEY is missing from .env")

llm = ChatOpenRouter(
    model="deepseek/deepseek-v4-flash-0731",
    api_key=api_key,
    temperature=0,
    max_tokens=500,
)
```

## **FIRST CHAIN**

```python
#TASK-1 prompt

from langchain_core.prompts import ChatPromptTemplate

prompt_template_1 = ChatPromptTemplate([
    ("system","You are an all in one assistant and can help with any task or question"),
    ("human","{input}")
])
```

```python
#TASK-2 llm

llm = ChatOpenRouter(
    model="deepseek/deepseek-v4-flash-0731",
    api_key=api_key,
    temperature=0,
    max_tokens=500,
)
```

```python
#TASK-3 String Parser

from langchain_core.output_parsers import StrOutputParser

str_parser = StrOutputParser()
```

#### **MANUAL STEPS WE LEARNT TILL NOW**

```python
template = prompt_template_1.invoke({"input":"Why do humans breathe?"})
result = llm.invoke(template)
result.content
```

#### **CHAIN INVOCATION**

```python
chain = prompt_template_1 | llm | str_parser

chain.invoke({"input":"What is bread?"})
```

```python
#SAME Thing with another function
from langchain_core.runnables import RunnableSequence

chain_same = RunnableSequence(prompt_template_1,llm,str_parser)

chain_same.invoke({"input":"What is bread?"})
```
