## **Boiler Plate**

```python
import os
from dotenv import load_dotenv
from langchain_openrouter import ChatOpenRouter

load_dotenv(dotenv_path="../.env",override=True)

api_key=os.getenv("OPENROUTER_API_KEY")
if not api_key:
    raise RuntimeError("OPENROUTER_API_KEY is missing from .env")

llm = ChatOpenRouter(
    model="deepseek/deepseek-v4-flash-0731",
    api_key=api_key,
    temperature=1.2,
    max_tokens=500,
)
```

## **Guiding in Prompts**

```python
llm.invoke("Tell me a joke. give output in JSON format with setup and punchline Key value pair").content
```

### **PYDANTIC**

```python
from pydantic import BaseModel,Field

class llm_schema(BaseModel):
    setup: str = Field(description="The setup for the joke")
    punchline: str = Field(description="The punchlinef for the joke")
```

```python
llm_structured_output = llm.with_structured_output(llm_schema)

llm_structured_output.invoke("Tell me a joke")
```
